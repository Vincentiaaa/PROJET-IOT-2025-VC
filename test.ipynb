```python
import streamlit as st
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
ds3 = pd.read_csv("DS3_weapon.csv")

del ds3["Sell Price"]
del ds3["Critical"]
del ds3["Stability"]

ds3.head()
```

```python
bestfire = ds3.sort_values(by="Fire Damage", ascending=False).iloc[:10]
st.write(bestfire.loc[:, ["Name", "Fire Damage"]])

fig, ax = plt.subplots()

ax.plot(bestfire["Name"], bestfire["Fire Damage"])
ax.set_xlabel("Name")
ax.set_ylabel("Fire Damage")
st.pyplot(fig)
```
